```python
print(df.columns)
```

```python
# Drop irrelevant columns
df=df.drop(columns=['Id','Name','Franchise','Category','City','No_Of_Item'])
```

```python
df.head()
```

```python
df.shape
```

```python
# Create independent feature and dependent variable
x=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train)
```

```python
print(y_train)
```

```python
print(y_test)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
# Save trained model for deployment
import pickle

with open('model.pkl', 'wb') as model_file:
    pickle.dump(model, model_file)

print('model.pkl saved successfully')
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
# Load saved model and test deployment prediction
with open('model.pkl', 'rb') as model_file:
    loaded_model = pickle.load(model_file)

# Example: predict revenue using the same input feature used during training
sample_input = [[100]]
print('Predicted Revenue:', loaded_model.predict(sample_input)[0])
```

```python
# Visualising the Training set results
plt.scatter(x_train,y_train,color="blue")  # scatter plot on x_train,y_train
plt.plot(x_train,model.predict(x_train),color="green")   # line plot on x_train
plt.title("Revenue vs orders")
plt.xlabel("orders")
plt.ylabel("Revenue")

```

```python
# Visualising the Test set results
plt.scatter(x_test,y_test,color="black")  # scatter plot on x_test,y_test
plt.plot(x_train,model.predict(x_train),color="red") # line plot on x_train
plt.title("Revenue vs orders")
plt.xlabel("orders")
plt.ylabel("Revenue")
```

```python
from sklearn.metrics import r2_score
R2_score=r2_score(y_pred,y_test)
print("R2 SCORE:",R2_score)
```
