# Restaurant Revenue Prediction — 3 Feature Model

This notebook retrains the model using exactly three input features for Streamlit deployment.

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
# Load dataset
df = pd.read_csv("revenue_prediction.csv")
df.head()
```

```python
# Select exactly the 3 features used by the Streamlit app
X = df[["No_Of_Item", "Order_Placed", "Id"]]
y = df["Revenue"]

print("Features:", X.columns.tolist())
print("Shape of X:", X.shape)
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
# Train Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Evaluate model
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("R2 Score:", r2)
print("MAE:", mae)
print("RMSE:", rmse)
```

```python
# Save the trained model
joblib.dump(model, "model.pkl")
print("model.pkl created successfully.")
```

```python
# Verify that the saved model expects exactly 3 features
loaded_model = joblib.load("model.pkl")
print("Number of input features:", loaded_model.n_features_in_)
print("Expected features: No_Of_Item, Order_Placed, Id")
```

```python
# Example prediction
# Number of Items = 50
# Orders Placed = 100
# Restaurant ID = 110

sample = np.array([[50, 100, 110]])
prediction = loaded_model.predict(sample)

print("Predicted Revenue:", prediction[0])
```
